## Part A: Main Experiment Implementation


```python
from itertools import permutations

INF = float('inf')
cost = [
    [INF, 10, 8, 9, 7],
    [10, INF, 10, 5, 6],
    [8, 10, INF, 8, 9],
    [9, 5, 8, INF, 6],
    [7, 6, 9, 6, INF]
]
cities = ['A', 'B', 'C', 'D', 'E']
best_cost, best_path = INF, None

for perm in permutations(range(1, 5)):
    path = [0] + list(perm) + [0]
    c = sum(cost[path[i]][path[i+1]] for i in range(5))
    if c < best_cost:
        best_cost, best_path = c, path

print("5-City TSP - Cost Matrix:")
print("     A    B    C    D    E")
for i, row in enumerate(cost):
    r = ['INF' if x == INF else str(x) for x in row]
    print(f"{cities[i]:>4}", ' '.join(f"{v:>4}" for v in r))

print(f"\nOptimal Tour: {' -> '.join([cities[i] for i in best_path])}")
print(f"Minimum Cost: {best_cost}")
```
