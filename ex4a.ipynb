## Part A: Main Experiment Implementation


```python
import heapq

def dijkstra(graph, source):
    n = len(graph)
    dist, prev = [float('inf')] * n, [None] * n
    dist[source] = 0
    pq, visited = [(0, source)], set()
    while pq:
        d, u = heapq.heappop(pq)
        if u in visited: continue
        visited.add(u)
        for v, w in graph[u]:
            if dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                prev[v] = u
                heapq.heappush(pq, (dist[v], v))
    return dist, prev

def reconstruct_path(prev, source, target):
    path, node = [], target
    while node is not None:
        path.append(node)
        node = prev[node]
    path.reverse()
    return path if path[0] == source else []

graph = {0: [(1, 4), (2, 1)], 1: [(3, 1)], 2: [(1, 2), (3, 5)], 3: [(4, 3)], 4: [(5, 2)], 5: []}
dist, prev = dijkstra(graph, 0)
print(f"Shortest paths from vertex 0:")
print(f"{'Vertex':>8} {'Distance':>10} {'Path':>30}")
print('-'*55)
for v in range(len(graph)):
    p = reconstruct_path(prev, 0, v)
    p_str = ' -> '.join(map(str, p)) if p else 'No path'
    print(f"{v:>8} {dist[v]:>10} {p_str:>30}")
```

```python
# Practice Problem 1: City Road Network (Airport to Station)
city_graph = {
    0: [(1, 10), (2, 5)],
    1: [(3, 1), (2, 2)],
    2: [(1, 3), (3, 9), (4, 2)],
    3: [(4, 4)],
    4: []
}
c_dist, c_prev = dijkstra(city_graph, 0)
c_path = reconstruct_path(c_prev, 0, 4)
print("--- Practice Problem 1: Airport to Central Station ---")
print(f"Fastest Route Airport (0) -> Central Station (4): {' -> '.join(map(str, c_path))}")
print(f"Total Travel Time: {c_dist[4]} minutes")

# Practice Problem 2: Modified Dijkstra with Hop Count
def dijkstra_hops(graph, source):
    n = len(graph)
    dist, hops, prev = [float('inf')] * n, [float('inf')] * n, [None] * n
    dist[source], hops[source] = 0, 0
    pq = [(0, 0, source)]
    while pq:
        d, h, u = heapq.heappop(pq)
        for v, w in graph[u]:
            if dist[u] + w < dist[v] or (dist[u] + w == dist[v] and hops[u] + 1 < hops[v]):
                dist[v] = dist[u] + w
                hops[v] = hops[u] + 1
                prev[v] = u
                heapq.heappush(pq, (dist[v], hops[v], v))
    return dist, hops, prev

d_h, h_h, p_h = dijkstra_hops(graph, 0)
print("\n--- Practice Problem 2: Shortest Distance with Hop Preference ---")
print(f"Distance to Node 5: {d_h[5]} | Hop Count: {h_h[5]} hops")
```
