## Part A: Main Experiment Implementation


```python
def matrix_chain_order(dims):
    n = len(dims) - 1
    m = [[0] * (n + 1) for _ in range(n + 1)]
    s = [[0] * (n + 1) for _ in range(n + 1)]
    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            m[i][j] = float('inf')
            for k in range(i, j):
                cost = m[i][k] + m[k + 1][j] + dims[i - 1] * dims[k] * dims[j]
                if cost < m[i][j]:
                    m[i][j] = cost; s[i][j] = k
    return m, s

def print_optimal_parens(s, i, j):
    if i == j: return f"A{i}"
    return f"({print_optimal_parens(s, i, s[i][j])} x {print_optimal_parens(s, s[i][j] + 1, j)})"

dims = [10, 30, 5, 60, 10]
m, s = matrix_chain_order(dims)
print("Matrix Dimensions:")
print("  A1: 10 x 30\n  A2: 30 x 5\n  A3: 5 x 60\n  A4: 60 x 10")
print(f"\nMinimum scalar multiplications: {m[1][4]}")
print(f"Optimal parenthesization: {print_optimal_parens(s, 1, 4)}")
```

## Part B & C: Practice Problems Solutions


```python
# Practice Problem 1: 5-Matrix Graphics Pipeline
gfx_dims = [4, 4, 4, 4, 10000, 1]
m_gfx, s_gfx = matrix_chain_order(gfx_dims)
print("--- Practice Problem 1: Graphics Pipeline Matrix Multiplication ---")
print(f"Optimal Scalar Multiplications: {m_gfx[1][5]}")
print(f"Optimal Parenthesization: {print_optimal_parens(s_gfx, 1, 5)}")

# Practice Problem 2: Top-Down Memoized MCM
memo = {}
def memoized_mcm(i, j, p):
    if i == j: return 0
    if (i, j) in memo: return memo[(i, j)]
    min_c = float('inf')
    for k in range(i, j):
        c = memoized_mcm(i, k, p) + memoized_mcm(k + 1, j, p) + p[i-1] * p[k] * p[j]
        min_c = min(min_c, c)
    memo[(i, j)] = min_c
    return min_c

print("\n--- Practice Problem 2: Top-Down Memoized MCM ---")
print(f"Top-down Memoized Minimum Cost: {memoized_mcm(1, 4, dims)}")
```
