```python
import heapq

def dijkstra_min_hops(graph, source):
    n = len(graph)
    dist = [float('inf')] * n
    hops = [float('inf')] * n
    dist[source], hops[source] = 0, 0
    pq = [(0, 0, source)]
    
    while pq:
        d, h, u = heapq.heappop(pq)
        for v, w in graph[u]:
            if dist[u] + w < dist[v] or (dist[u] + w == dist[v] and hops[u] + 1 < hops[v]):
                dist[v] = dist[u] + w
                hops[v] = hops[u] + 1
                heapq.heappush(pq, (dist[v], hops[v], v))
    return dist, hops

graph = {
    0: [(1, 4), (2, 1)],
    1: [(3, 1)],
    2: [(1, 2), (3, 5)],
    3: [(4, 3)],
    4: [(5, 2)],
    5: []
}

dist, hops = dijkstra_min_hops(graph, 0)

print("--- Practice Problem 4c: Dijkstra with Min-Hop Tie-Breaking ---")
for v in range(len(graph)):
    print(f"Destination Node {v}: Distance = {dist[v]}, Hop Count = {hops[v]}")
```
