```python
import random

def interpolation_search_float(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0
    while low <= high and arr[low] <= target <= arr[high]:
        comparisons += 1
        if arr[high] == arr[low]:
            return (low, comparisons) if arr[low] == target else (-1, comparisons)
        pos = low + int(((target - arr[low]) * (high - low)) / (arr[high] - arr[low]))
        if pos < low or pos > high:
            break
        if abs(arr[pos] - target) < 1e-6:
            return pos, comparisons
        elif arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1
    return -1, comparisons

print("--- Practice Problem 1c: Uniform Floating-Point Search Analysis ---")
print(f"{'Dataset Size':>12} | {'Comparisons':>15}")
print("-" * 32)

for size in [10000, 50000, 100000]:
    float_arr = sorted([random.uniform(0.0, 1000.0) for _ in range(size)])
    target_fl = float_arr[size // 2]
    _, comps = interpolation_search_float(float_arr, target_fl)
    print(f"{size:>12} | {comps:>15}")
```
