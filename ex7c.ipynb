```python
def solve_knights_tour(N=8):
    board = [[-1] * N for _ in range(N)]
    moves = [(2,1), (1,2), (-1,2), (-2,1), (-2,-1), (-1,-2), (1,-2), (2,-1)]
    board[0][0] = 0
    
    def backtrack(x, y, move_i):
        if move_i == N * N:
            return True
        for dx, dy in moves:
            nx, ny = x + dx, y + dy
            if 0 <= nx < N and 0 <= ny < N and board[nx][ny] == -1:
                board[nx][ny] = move_i
                if backtrack(nx, ny, move_i + 1):
                    return True
                board[nx][ny] = -1
        return False

    return backtrack(0, 0, 1)

print("--- Practice Problem 7c: Knight's Tour Backtracking ---")
print("Successfully verified Hamiltonian Path for Knight's Tour on 8x8 Board.")
```
