```python
def naive_2d_search(text_matrix, pattern_matrix):
    R, C = len(text_matrix), len(text_matrix[0])
    r, c = len(pattern_matrix), len(pattern_matrix[0])
    comparisons = 0
    matches = []

    for i in range(R - r + 1):
        for j in range(C - c + 1):
            match = True
            for x in range(r):
                for y in range(c):
                    comparisons += 1
                    if text_matrix[i + x][j + y] != pattern_matrix[x][y]:
                        match = False
                        break
                if not match:
                    break
            if match:
                matches.append((i, j))
    return matches, comparisons

# 10x10 text matrix and 3x3 pattern matrix
text_grid = [['A' for _ in range(10)] for _ in range(10)]
pattern_grid = [['A' for _ in range(3)] for _ in range(3)]

matches, comps = naive_2d_search(text_grid, pattern_grid)

print("--- Practice Problem 2c: 2D Matrix Pattern Matching ---")
print(f"Text Grid Size: 10x10 | Pattern Grid Size: 3x3")
print(f"Total Top-Left Matches Found: {len(matches)}")
print(f"Total Element Comparisons Required: {comps}")
```
