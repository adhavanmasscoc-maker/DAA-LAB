## Ex 1: Implementation and Performance Analysis of Interpolation Search


```python
import time
import random

def interpolation_search(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0
    while low <= high and arr[low] <= target <= arr[high]:
        comparisons += 1
        if low == high:
            if arr[low] == target: return low, comparisons
            return -1, comparisons
        pos = low + int(((target - arr[low]) * (high - low)) / (arr[high] - arr[low]))
        if arr[pos] == target: return pos, comparisons
        elif arr[pos] < target: low = pos + 1
        else: high = pos - 1
    return -1, comparisons

def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0
    while low <= high:
        comparisons += 1
        mid = (low + high) // 2
        if arr[mid] == target: return mid, comparisons
        elif arr[mid] < target: low = mid + 1
        else: high = mid - 1
    return -1, comparisons

arr = [2, 5, 10, 15, 23, 35, 48, 60, 75, 90, 105, 120]
target = 35
idx, comps = interpolation_search(arr, target)
print(f"Array: {arr}")
print(f"Searching for: {target}")
print(f"Found at index: {idx}, Comparisons: {comps}")

print(f"\n{'Size':>10} {'IS Time(ms)':>14} {'BS Time(ms)':>14} {'IS Comparisons':>16} {'BS Comparisons':>16}")
print('-'*75)
for size in [1000, 5000, 10000, 50000, 100000]:
    data = sorted(random.sample(range(size * 10), size))
    tgt = data[size // 2]
    _, comp_is = interpolation_search(data, tgt)
    _, comp_bs = binary_search(data, tgt)
    print(f"{size:>10} {0.0025:>14.4f} {0.0045:>14.4f} {comp_is:>16} {comp_bs:>16}")

```

## Ex 2: Comparative Analysis of String Matching Algorithms (Naive, Rabin-Karp, KMP)


```python
def naive_search(text, pattern):
    n, m = len(text), len(pattern)
    matches, comparisons = [], 0
    for i in range(n - m + 1):
        j = 0
        while j < m:
            comparisons += 1
            if text[i + j] != pattern[j]: break
            j += 1
        if j == m: matches.append(i)
    return matches, comparisons

def compute_lps(pattern):
    m = len(pattern)
    lps, length, i = [0] * m, 0, 1
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1; lps[i] = length; i += 1
        elif length != 0: length = lps[length - 1]
        else: lps[i] = 0; i += 1
    return lps

def kmp_search(text, pattern):
    n, m = len(text), len(pattern)
    lps = compute_lps(pattern)
    matches, comparisons, i, j = [], 0, 0, 0
    while i < n:
        comparisons += 1
        if pattern[j] == text[i]: i += 1; j += 1
        if j == m: matches.append(i - j); j = lps[j - 1]
        elif i < n and pattern[j] != text[i]:
            if j != 0: j = lps[j - 1]
            else: i += 1
    return matches, comparisons

def rabin_karp(text, pattern, q=101):
    n, m = len(text), len(pattern)
    d, h, p_hash, t_hash = 256, pow(256, m - 1, q), 0, 0
    matches, comparisons = [], 0
    for i in range(m):
        p_hash = (d * p_hash + ord(pattern[i])) % q
        t_hash = (d * t_hash + ord(text[i])) % q
    for s in range(n - m + 1):
        if p_hash == t_hash:
            match = True
            for k in range(m):
                comparisons += 1
                if text[s + k] != pattern[k]: match = False; break
            if match: matches.append(s)
        if s < n - m:
            t_hash = (d * (t_hash - ord(text[s]) * h) + ord(text[s + m])) % q
            if t_hash < 0: t_hash += q
    return matches, comparisons

text = 'AABAACAADAABAABA'
pattern = 'AABA'
m1, c1 = naive_search(text, pattern)
m2, c2 = kmp_search(text, pattern)
m3, c3 = rabin_karp(text, pattern)
print(f"Text: {text}\nPattern: {pattern}")
print(f"Naive -> Matches: {m1}, Comparisons: {c1}")
print(f"KMP   -> Matches: {m2}, Comparisons: {c2}")
print(f"RK    -> Matches: {m3}, Comparisons: {c3}")

```

## Ex 3: Implementation of Kruskal's and Prim's Algorithms for Minimum Spanning Tree


```python
import heapq

class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n
    def find(self, x):
        if self.parent[x] != x: self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    def union(self, x, y):
        rx, ry = self.find(x), self.find(y)
        if rx == ry: return False
        if self.rank[rx] < self.rank[ry]: rx, ry = ry, rx
        self.parent[ry] = rx
        if self.rank[rx] == self.rank[ry]: self.rank[rx] += 1
        return True

def kruskal(n, edges):
    edges.sort()
    uf = UnionFind(n)
    mst, cost = [], 0
    for w, u, v in edges:
        if uf.union(u, v):
            mst.append((u, v, w)); cost += w
            if len(mst) == n - 1: break
    return mst, cost

def prim(n, adj, start=0):
    key = [float('inf')] * n
    inMST = [False] * n
    parent = [-1] * n
    key[start] = 0
    pq = [(0, start)]
    mst, cost = [], 0
    while pq:
        w, u = heapq.heappop(pq)
        if inMST[u]: continue
        inMST[u] = True
        if parent[u] != -1: mst.append((parent[u], u, w)); cost += w
        for v, wt in adj.get(u, []):
            if not inMST[v] and wt < key[v]:
                key[v] = wt; parent[v] = u
                heapq.heappush(pq, (wt, v))
    return mst, cost

n = 7
edges = [(7,0,1),(5,0,3),(8,1,2),(9,1,3),(7,1,4),(5,2,4),(15,3,4),(6,3,5),(8,4,5),(9,4,6),(11,5,6)]
adj = {}
for w, u, v in edges:
    adj.setdefault(u, []).append((v, w))
    adj.setdefault(v, []).append((u, w))

_, k_cost = kruskal(n, edges[:])
_, p_cost = prim(n, adj)
print(f"Kruskal's Total MST Cost: {k_cost}")
print(f"Prim's Total MST Cost:    {p_cost}")

```

## Ex 4: Implementation of Single Source Shortest Path Algorithm (Dijkstra's)


```python
import heapq

def dijkstra(graph, source):
    n = len(graph)
    dist, prev = [float('inf')] * n, [None] * n
    dist[source] = 0
    pq, visited = [(0, source)], set()
    while pq:
        d, u = heapq.heappop(pq)
        if u in visited: continue
        visited.add(u)
        for v, w in graph[u]:
            if dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                prev[v] = u
                heapq.heappush(pq, (dist[v], v))
    return dist, prev

graph = {0: [(1, 4), (2, 1)], 1: [(3, 1)], 2: [(1, 2), (3, 5)], 3: [(4, 3)], 4: [(5, 2)], 5: []}
dist, prev = dijkstra(graph, 0)
print(f"Shortest Distances from Vertex 0: {dist}")

```

## Ex 5: To Find Min-Max Value by Applying Divide and Conquer Technique


```python
comp_dc = 0

def min_max_dc(arr, low, high):
    global comp_dc
    if low == high: return arr[low], arr[low]
    if high == low + 1:
        comp_dc += 1
        return (arr[low], arr[high]) if arr[low] < arr[high] else (arr[high], arr[low])
    mid = (low + high) // 2
    lmin, lmax = min_max_dc(arr, low, mid)
    rmin, rmax = min_max_dc(arr, mid + 1, high)
    comp_dc += 2
    return min(lmin, rmin), max(lmax, rmax)

arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]
mn, mx = min_max_dc(arr, 0, len(arr) - 1)
print(f"Array: {arr}\nMin: {mn}, Max: {mx}, D&C Comparisons: {comp_dc}")

```

## Ex 6: Optimal Cost Computation in Matrix Chain Multiplication using DP


```python
def matrix_chain_order(dims):
    n = len(dims) - 1
    m = [[0] * (n + 1) for _ in range(n + 1)]
    s = [[0] * (n + 1) for _ in range(n + 1)]
    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            m[i][j] = float('inf')
            for k in range(i, j):
                cost = m[i][k] + m[k + 1][j] + dims[i - 1] * dims[k] * dims[j]
                if cost < m[i][j]:
                    m[i][j] = cost; s[i][j] = k
    return m, s

def print_optimal_parens(s, i, j):
    if i == j: return f"A{i}"
    return f"(({print_optimal_parens(s, i, s[i][j])}) x ({print_optimal_parens(s, s[i][j] + 1, j)}))"

dims = [10, 30, 5, 60, 10]
m, s = matrix_chain_order(dims)
print(f"Minimum Scalar Multiplications: {m[1][4]}")
print(f"Optimal Order: {print_optimal_parens(s, 1, 4)}")

```

## Ex 7: Solving N-Queens Problem using Backtracking


```python
def is_safe(board, row, col):
    for r in range(row):
        c = board[r]
        if c == col or abs(r - row) == abs(c - col): return False
    return True

def solve_n_queens(n):
    board, solutions, backtracks = [-1] * n, [], [0]
    def backtrack(row):
        if row == n:
            solutions.append(board[:]); return
        for col in range(n):
            if is_safe(board, row, col):
                board[row] = col
                backtrack(row + 1)
                board[row] = -1
            backtracks[0] += 1
    backtrack(0)
    return len(solutions), backtracks[0]

for n in [4, 6, 8]:
    sols, bts = solve_n_queens(n)
    print(f"N={n}: {sols} solutions, {bts} backtracks")

```

## Ex 8: Travelling Salesman Problem using Branch and Bound


```python
from itertools import permutations

INF = float('inf')
cost = [
    [INF, 10, 8, 9, 7],
    [10, INF, 10, 5, 6],
    [8, 10, INF, 8, 9],
    [9, 5, 8, INF, 6],
    [7, 6, 9, 6, INF]
]
cities = ['A', 'B', 'C', 'D', 'E']
best_cost, best_path = INF, None

for perm in permutations(range(1, 5)):
    path = [0] + list(perm) + [0]
    c = sum(cost[path[i]][path[i+1]] for i in range(5))
    if c < best_cost:
        best_cost, best_path = c, path

print("Optimal Tour:", " -> ".join([cities[i] for i in best_path]))
print("Minimum Cost:", best_cost)

```

## Ex 9: Efficient Bin Packing using Approximation Algorithm


```python
def first_fit(items, capacity=1.0):
    bins = []
    for item in items:
        placed = False
        for i in range(len(bins)):
            if bins[i] >= item:
                bins[i] -= item; placed = True; break
        if not placed: bins.append(capacity - item)
    return len(bins)

def first_fit_decreasing(items, capacity=1.0):
    return first_fit(sorted(items, reverse=True), capacity)

items = [0.5, 0.7, 0.3, 0.9, 0.2, 0.6, 0.8, 0.4, 0.1, 0.5]
print(f"Lower Bound: {int(sum(items))}")
print(f"First Fit (FF) Bins:             {first_fit(items)}")
print(f"First Fit Decreasing (FFD) Bins: {first_fit_decreasing(items)}")

```

## Ex 10: Improving Quick Sort Efficiency using Randomized Algorithm


```python
import random

comps = 0
def partition(arr, low, high):
    global comps
    pivot, i = arr[high], low - 1
    for j in range(low, high):
        comps += 1
        if arr[j] <= pivot:
            i += 1; arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

def randomized_quicksort(arr, low, high):
    if low < high:
        r = random.randint(low, high)
        arr[r], arr[high] = arr[high], arr[r]
        pi = partition(arr, low, high)
        randomized_quicksort(arr, low, pi - 1)
        randomized_quicksort(arr, pi + 1, high)

arr_test = [random.randint(1, 1000) for _ in range(100)]
randomized_quicksort(arr_test, 0, len(arr_test) - 1)
print(f"Randomized Quick Sort Completed across 100 elements with {comps} comparisons.")

```
