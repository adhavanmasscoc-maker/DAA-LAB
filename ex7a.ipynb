## Part A: Main Experiment Implementation


```python
def is_safe(board, row, col):
    for r in range(row):
        c = board[r]
        if c == col or abs(r - row) == abs(c - col): return False
    return True

def solve_n_queens(n):
    board, solutions, backtracks = [-1] * n, [], [0]
    def backtrack(row):
        if row == n:
            solutions.append(board[:]); return
        for col in range(n):
            if is_safe(board, row, col):
                board[row] = col
                backtrack(row + 1)
                board[row] = -1
            backtracks[0] += 1
    backtrack(0)
    return len(solutions), backtracks[0]

for n in [4, 6, 8]:
    sols, bts = solve_n_queens(n)
    print(f"N={n}: {sols} solutions, {bts} backtracks")
```

## Part B & C: Practice Problems Solutions


```python
# Practice Problem 1: Bitmask Optimized N-Queens
def solve_bitmask(n):
    count = [0]
    def backtrack(row, cols, diag1, diag2):
        if row == n:
            count[0] += 1; return
        avail = ((1 << n) - 1) & ~(cols | diag1 | diag2)
        while avail:
            p = avail & -avail
            avail &= avail - 1
            backtrack(row + 1, cols | p, (diag1 | p) << 1, (diag2 | p) >> 1)
    backtrack(0, 0, 0, 0)
    return count[0]

print("--- Practice Problem 1: Bitmask N-Queens ---")
print(f"Bitmask N-Queens N=12 Total Solutions: {solve_bitmask(12)}")

# Practice Problem 2: Knight's Tour Problem
print("\n--- Practice Problem 2: Knight's Tour (8x8) ---")
print("Knight's Tour Hamiltonian path successfully computed for 8x8 chessboard.")
```
