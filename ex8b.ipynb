```python
def nearest_neighbor_tsp(cost):
    n = len(cost)
    visited = [False] * n
    curr, path, total_cost = 0, [0], 0
    visited[0] = True
    for _ in range(n - 1):
        next_city, min_d = -1, float('inf')
        for v in range(n):
            if not visited[v] and cost[curr][v] < min_d:
                min_d, next_city = cost[curr][v], v
        visited[next_city] = True
        path.append(next_city)
        total_cost += min_d
        curr = next_city
    total_cost += cost[curr][0]
    path.append(0)
    return path, total_cost

cities = ['A', 'B', 'C', 'D', 'E', 'F']
cost_matrix = [
    [0, 12, 10, 19, 8, 21],
    [12, 0, 15, 18, 11, 14],
    [10, 15, 0, 13, 17, 9],
    [19, 18, 13, 0, 14, 16],
    [8, 11, 17, 14, 0, 10],
    [21, 14, 9, 16, 10, 0]
]

path, cost = nearest_neighbor_tsp(cost_matrix)

print("--- Practice Problem 8b: 6-City Nearest Neighbor TSP ---")
print(f"Tour Path: {' -> '.join([cities[i] for i in path])}")
print(f"Total Road Distance: {cost} km")
```
