```python
import heapq

def dijkstra(graph, source):
    n = len(graph)
    dist = [float('inf')] * n
    prev = [None] * n
    dist[source] = 0
    pq = [(0, source)]
    while pq:
        d, u = heapq.heappop(pq)
        if d > dist[u]: continue
        for v, w in graph[u]:
            if dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                prev[v] = u
                heapq.heappush(pq, (dist[v], v))
    return dist, prev

city_graph = {
    0: [(1, 10), (2, 5)],
    1: [(3, 1), (2, 2)],
    2: [(1, 3), (3, 9), (4, 2)],
    3: [(4, 4)],
    4: []
}

dist, prev = dijkstra(city_graph, 0)

path, curr = [], 4
while curr is not None:
    path.append(curr)
    curr = prev[curr]
path.reverse()

print("--- Practice Problem 4b: Airport to Central Station Route ---")
print(f"Fastest Path: {' -> '.join(map(str, path))}")
print(f"Total Minimum Travel Time: {dist[4]} minutes")
```
