```python
def solve_n_queens_bitmask(n):
    count = [0]
    def backtrack(row, cols, pos_diag, neg_diag):
        if row == n:
            count[0] += 1
            return
        valid_positions = ((1 << n) - 1) & ~(cols | pos_diag | neg_diag)
        while valid_positions:
            bit = valid_positions & -valid_positions
            valid_positions ^= bit
            backtrack(row + 1, cols | bit, (pos_diag | bit) << 1, (neg_diag | bit) >> 1)
            
    backtrack(0, 0, 0, 0)
    return count[0]

ans_12 = solve_n_queens_bitmask(12)

print("--- Practice Problem 7b: Bitmask Optimized N-Queens ---")
print(f"Total Valid Solutions for N=12 Queens: {ans_12}")
```
