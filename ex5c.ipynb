# Ex 5c: QuickSelect for K-th Smallest Element

```python
import random

def quickselect(arr, k):
    if not arr: return None
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    mid = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    if k <= len(left):
        return quickselect(left, k)
    elif k <= len(left) + len(mid):
        return pivot
    else:
        return quickselect(right, k - len(left) - len(mid))

sample = list(range(1, 1001))
random.shuffle(sample)

print("--- Practice Problem 5c: QuickSelect k-th Smallest Element ---")
print(f"k=1    (Minimum): {quickselect(sample, 1)}")
print(f"k=500  (Median) : {quickselect(sample, 500)}")
print(f"k=1000 (Maximum): {quickselect(sample, 1000)}")
```
