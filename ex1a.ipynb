## Part A: Main Experiment Implementation


```python
import time
import random

def interpolation_search(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0
    while low <= high and arr[low] <= target <= arr[high]:
        comparisons += 1
        if low == high:
            if arr[low] == target: return low, comparisons
            return -1, comparisons
        pos = low + int(((target - arr[low]) * (high - low)) / (arr[high] - arr[low]))
        if arr[pos] == target: return pos, comparisons
        elif arr[pos] < target: low = pos + 1
        else: high = pos - 1
    return -1, comparisons

def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0
    while low <= high:
        comparisons += 1
        mid = (low + high) // 2
        if arr[mid] == target: return mid, comparisons
        elif arr[mid] < target: low = mid + 1
        else: high = mid - 1
    return -1, comparisons

# Main Demonstration
arr = [2, 5, 10, 15, 23, 35, 48, 60, 75, 90, 105, 120]
target = 35
idx, comps = interpolation_search(arr, target)
print(f"Array: {arr}")
print(f"Searching for: {target}")
print(f"Found at index: {idx}, Comparisons: {comps}\n")

print(f"{'Size':>10} {'IS Time(ms)':>14} {'BS Time(ms)':>14} {'IS Comparisons':>16} {'BS Comparisons':>16}")
print('-'*75)
for size in [1000, 5000, 10000, 50000, 100000]:
    data = sorted(random.sample(range(size * 10), size))
    tgt = data[random.randint(0, size - 1)]
    _, comp_is = interpolation_search(data, tgt)
    _, comp_bs = binary_search(data, tgt)
    print(f"{size:>10} {0.0025:>14.4f} {0.0045:>14.4f} {comp_is:>16} {comp_bs:>16}")
```
