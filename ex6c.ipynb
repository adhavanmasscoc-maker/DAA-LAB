```python
memo = {}

def memoized_mcm(i, j, p):
    if i == j: return 0
    if (i, j) in memo: return memo[(i, j)]
    min_cost = float('inf')
    for k in range(i, j):
        cost = memoized_mcm(i, k, p) + memoized_mcm(k + 1, j, p) + p[i-1] * p[k] * p[j]
        min_cost = min(min_cost, cost)
    memo[(i, j)] = min_cost
    return min_cost

dims = [10, 30, 5, 60, 10]
cost = memoized_mcm(1, 4, dims)

print("--- Practice Problem 6c: Top-Down Memoized MCM ---")
print(f"Dimensions: {dims}")
print(f"Minimum Scalar Multiplications (Memoized): {cost}")
```
