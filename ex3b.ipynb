```python
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
    def find(self, i):
        if self.parent[i] == i: return i
        self.parent[i] = self.find(self.parent[i])
        return self.parent[i]
    def union(self, i, j):
        root_i, root_j = self.find(i), self.find(j)
        if root_i != root_j:
            self.parent[root_i] = root_j
            return True
        return False

cities = ['A', 'B', 'C', 'D', 'E', 'F']
edges = [
    (4, 0, 1), (2, 0, 2), (5, 1, 2), (10, 1, 3),
    (3, 2, 4), (7, 3, 4), (8, 3, 5), (6, 4, 5)
]

edges.sort()
uf = UnionFind(6)
mst_edges, total_cost = [], 0

for w, u, v in edges:
    if uf.union(u, v):
        mst_edges.append((cities[u], cities[v], w))
        total_cost += w

print("--- Practice Problem 3b: 6-City MST via Kruskal's ---")
for u, v, w in mst_edges:
    print(f"  Connection: {u} <---> {v} | Cost: {w}")
print(f"Total Minimum Connection Cost: {total_cost}")
```
