```python
def two_opt(path, cost):
    best_path = path[:]
    improved = True
    def tour_cost(p):
        return sum(cost[p[i]][p[i+1]] for i in range(len(p)-1))
    best_cost = tour_cost(best_path)
    while improved:
        improved = False
        for i in range(1, len(best_path) - 2):
            for j in range(i + 1, len(best_path) - 1):
                new_path = best_path[:i] + best_path[i:j+1][::-1] + best_path[j+1:]
                new_cost = tour_cost(new_path)
                if new_cost < best_cost:
                    best_path, best_cost = new_path, new_cost
                    improved = True
    return best_path, best_cost

cost_matrix = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

initial_path = [0, 1, 2, 3, 0]
opt_path, opt_cost = two_opt(initial_path, cost_matrix)

print("--- Practice Problem 8c: 2-Opt TSP Refinement ---")
print(f"Initial Path: {initial_path}")
print(f"Optimized Path: {opt_path} | Total Cost: {opt_cost}")
```
