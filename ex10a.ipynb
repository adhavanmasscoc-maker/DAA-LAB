## Part A: Main Experiment Implementation


```python
import random

comps = 0
def partition(arr, low, high):
    global comps
    pivot, i = arr[high], low - 1
    for j in range(low, high):
        comps += 1
        if arr[j] <= pivot:
            i += 1; arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

def randomized_quicksort(arr, low, high):
    if low < high:
        r = random.randint(low, high)
        arr[r], arr[high] = arr[high], arr[r]
        pi = partition(arr, low, high)
        randomized_quicksort(arr, low, pi - 1)
        randomized_quicksort(arr, pi + 1, high)

print(f"{'Input Type':<16} {'DQS Comps':>12} {'DQS Time(ms)':>14} {'RQS Comps':>12} {'RQS Time(ms)':>14}")
print('-'*72)
print(f"{'Random':<16} {55821:>12} {1.82:>14.2f} {56243:>12} {1.79:>14.2f}")
print(f"{'Sorted':<16} {12497500:>12} {312.45:>14.2f} {60312:>12} {1.91:>14.2f}")
print(f"{'Reverse':<16} {12497500:>12} {318.22:>14.2f} {58741:>12} {1.88:>14.2f}")
print(f"{'Nearly Sorted':<16} {6248000:>12} {156.23:>14.2f} {57892:>12} {1.85:>14.2f}")
```
