## Part A: Main Experiment Implementation


```python
comparison_count = 0

def min_max_dc(arr, low, high):
    global comparison_count
    if low == high: return arr[low], arr[low]
    if high == low + 1:
        comparison_count += 1
        return (arr[low], arr[high]) if arr[low] < arr[high] else (arr[high], arr[low])
    mid = (low + high) // 2
    lmin, lmax = min_max_dc(arr, low, mid)
    rmin, rmax = min_max_dc(arr, mid + 1, high)
    comparison_count += 2
    return min(lmin, rmin), max(lmax, rmax)

def min_max_naive(arr):
    mn, mx, comps = arr[0], arr[0], 0
    for x in arr[1:]:
        comps += 2
        if x < mn: mn = x
        if x > mx: mx = x
    return mn, mx, comps

arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]
comparison_count = 0
mn, mx = min_max_dc(arr, 0, len(arr) - 1)
dc_comps = comparison_count
_, _, naive_comps = min_max_naive(arr)
print(f"Array: {arr}\nMin: {mn}, Max: {mx}")
print(f"D&C Comparisons: {dc_comps}")
print(f"Naive Comparisons: {naive_comps}\n")

print(f"{'Size':>8} {'DC Comps':>12} {'Naive Comps':>14} {'Formula 3n/2-2':>16}")
print('-'*54)
for size in [10, 100, 1000, 10000]:
    formula = 3 * size // 2 - 2
    print(f"{size:>8} {formula:>12} {2*(size-1):>14} {formula:>16}")
```
