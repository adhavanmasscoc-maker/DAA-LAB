```python
def next_fit_decreasing(items, capacity=1.0):
    sorted_items = sorted(items, reverse=True)
    bins_count = 1
    rem_space = capacity
    for item in sorted_items:
        if item <= rem_space:
            rem_space -= item
        else:
            bins_count += 1
            rem_space = capacity - item
    return bins_count

items = [0.4, 0.8, 0.2, 0.1, 0.7, 0.5]
bins_used = next_fit_decreasing(items)

print("--- Practice Problem 9c: Next Fit Decreasing (NFD) ---")
print(f"Items: {items}")
print(f"Total Bins Used (NFD): {bins_used}")
```
