```python
def compute_lps(pattern):
    m = len(pattern)
    lps = [0] * m
    length, i = 0, 1
    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        elif length != 0:
            length = lps[length - 1]
        else:
            lps[i] = 0
            i += 1
    return lps

def kmp_search(text, pattern):
    n, m = len(text), len(pattern)
    lps = compute_lps(pattern)
    matches, comparisons, i, j = [], 0, 0, 0
    while i < n:
        comparisons += 1
        if pattern[j] == text[i]:
            i += 1
            j += 1
        if j == m:
            matches.append(i - j)
            j = lps[j - 1]
        elif i < n and pattern[j] != text[i]:
            if j != 0:
                j = lps[j - 1]
            else:
                i += 1
    return matches, comparisons, lps

text = "AABCAABXAAAZ"
pattern = "AABXAA"
matches, comps, lps = kmp_search(text, pattern)

print("--- Practice Problem 2b: KMP Trace & Verification ---")
print(f"Text:    {text}")
print(f"Pattern: {pattern}")
print(f"LPS Array: {lps}")
print(f"Pattern Matches found at index: {matches}")
print(f"Total Character Comparisons: {comps}")
```
