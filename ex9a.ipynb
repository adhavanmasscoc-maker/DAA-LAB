## Part A: Main Experiment Implementation


```python
def first_fit(items, capacity=1.0):
    bins = []
    for item in items:
        placed = False
        for i in range(len(bins)):
            if bins[i] >= item:
                bins[i] -= item; placed = True; break
        if not placed: bins.append(capacity - item)
    return len(bins)

def first_fit_decreasing(items, capacity=1.0):
    return first_fit(sorted(items, reverse=True), capacity)

def best_fit_decreasing(items, capacity=1.0):
    sorted_items = sorted(items, reverse=True)
    bins = []
    for item in sorted_items:
        best_idx, best_space = -1, float('inf')
        for i, space in enumerate(bins):
            if space >= item and space - item < best_space:
                best_space = space - item
                best_idx = i
        if best_idx >= 0: bins[best_idx] -= item
        else: bins.append(capacity - item)
    return len(bins)

items = [0.5, 0.7, 0.3, 0.9, 0.2, 0.6, 0.8, 0.4, 0.1, 0.5]
print(f"Items: {items}")
print(f"Capacity: 1.0")
print(f"Lower Bound on Bins: 5")
print(f"First Fit (FF) Bins:             {first_fit(items)}")
print(f"First Fit Decreasing (FFD) Bins: {first_fit_decreasing(items)}")
print(f"Best Fit Decreasing (BFD) Bins:  {best_fit_decreasing(items)}")
```
