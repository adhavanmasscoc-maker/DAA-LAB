```python
def matrix_chain_order(dims):
    n = len(dims) - 1
    m = [[0] * (n + 1) for _ in range(n + 1)]
    s = [[0] * (n + 1) for _ in range(n + 1)]
    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            m[i][j] = float('inf')
            for k in range(i, j):
                cost = m[i][k] + m[k + 1][j] + dims[i - 1] * dims[k] * dims[j]
                if cost < m[i][j]:
                    m[i][j] = cost
                    s[i][j] = k
    return m, s

def print_parens(s, i, j):
    if i == j: return f"M{i}"
    return f"({print_parens(s, i, s[i][j])} x {print_parens(s, s[i][j] + 1, j)})"

# Dimensions: T(4x4), R(4x4), S(4x4), V(4x10000), P(10000x1)
gfx_dims = [4, 4, 4, 4, 10000, 1]
m, s = matrix_chain_order(gfx_dims)

print("--- Practice Problem 6b: 3D Graphics Matrix Multiplication ---")
print(f"Minimum Scalar Multiplications: {m[1][5]}")
print(f"Optimal Order: {print_parens(s, 1, 5)}")
```
