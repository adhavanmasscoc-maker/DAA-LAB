```python
def second_min_max(arr):
    unique_sorted = sorted(list(set(arr)))
    if len(unique_sorted) < 2:
        return None, None
    return unique_sorted[1], unique_sorted[-2]

data = [12, 35, 1, 10, 34, 1, 35]
s_min, s_max = second_min_max(data)

print("--- Practice Problem 5b: Second Min & Second Max ---")
print(f"Input Array: {data}")
print(f"Second Minimum Element: {s_min}")
print(f"Second Maximum Element: {s_max}")
```
