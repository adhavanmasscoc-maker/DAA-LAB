## Part A: Main Experiment Implementation


```python
import heapq

class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n
    def find(self, x):
        if self.parent[x] != x: self.parent[x] = self.find(self.parent[x])
        return self.parent[x]
    def union(self, x, y):
        rx, ry = self.find(x), self.find(y)
        if rx == ry: return False
        if self.rank[rx] < self.rank[ry]: rx, ry = ry, rx
        self.parent[ry] = rx
        if self.rank[rx] == self.rank[ry]: self.rank[rx] += 1
        return True

def kruskal(n, edges):
    edges.sort()
    uf = UnionFind(n)
    mst, cost = [], 0
    for w, u, v in edges:
        if uf.union(u, v):
            mst.append((u, v, w)); cost += w
            if len(mst) == n - 1: break
    return mst, cost

def prim(n, adj, start=0):
    key = [float('inf')] * n
    inMST = [False] * n
    parent = [-1] * n
    key[start] = 0
    pq = [(0, start)]
    mst, cost = [], 0
    while pq:
        w, u = heapq.heappop(pq)
        if inMST[u]: continue
        inMST[u] = True
        if parent[u] != -1: mst.append((parent[u], u, w)); cost += w
        for v, wt in adj.get(u, []):
            if not inMST[v] and wt < key[v]:
                key[v] = wt; parent[v] = u
                heapq.heappush(pq, (wt, v))
    return mst, cost

n = 7
edges = [(7,0,1),(5,0,3),(8,1,2),(9,1,3),(7,1,4),(5,2,4),(15,3,4),(6,3,5),(8,4,5),(9,4,6),(11,5,6)]
adj = {}
for w, u, v in edges:
    adj.setdefault(u, []).append((v, w))
    adj.setdefault(v, []).append((u, w))

kmst, k_cost = kruskal(n, edges[:])
pmst, p_cost = prim(n, adj)
print(f"=== Kruskal's MST ===")
for u, v, w in kmst: print(f" Edge ({u} - {v}) Weight: {w}")
print(f" Total MST Cost: {k_cost}")
print(f"\n=== Prim's MST ===")
for u, v, w in pmst: print(f" Edge ({u} - {v}) Weight: {w}")
print(f" Total MST Cost: {p_cost}")
```

```python
# Practice Problem 1: 6-City Kruskal Algorithm
cities = ['A', 'B', 'C', 'D', 'E', 'F']
p1_edges = [(4,0,1), (2,0,2), (5,1,2), (10,1,3), (3,2,4), (7,3,4), (8,3,5), (6,4,5)]
mst, cost = kruskal(6, p1_edges)
print("--- Practice Problem 1: 6-City MST ---")
for u, v, w in mst:
    print(f"  {cities[u]} - {cities[v]} : {w}")
print(f"Total Minimum Spanning Tree Cost: {cost}")

# Practice Problem 2: Prim's Algorithm using Adjacency Matrix O(V^2)
def prim_matrix(n, matrix):
    key = [float('inf')] * n
    inMST = [False] * n
    key[0] = 0
    for _ in range(n):
        u = min((k, i) for i, k in enumerate(key) if not inMST[i])[1]
        inMST[u] = True
        for v in range(n):
            if matrix[u][v] > 0 and not inMST[v] and matrix[u][v] < key[v]:
                key[v] = matrix[u][v]
    return sum(key)

matrix = [[0 if i == j else (i + j) % 10 + 1 for j in range(100)] for i in range(100)]
print("\n--- Practice Problem 2: Dense Graph Adjacency Matrix Prim's ---")
print(f"Dense Graph (100 Vertices) Matrix Prim MST Cost: {prim_matrix(100, matrix)}")
```
