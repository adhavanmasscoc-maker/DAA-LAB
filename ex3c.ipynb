```python
def prim_matrix(n, matrix):
    key = [float('inf')] * n
    inMST = [False] * n
    key[0] = 0
    for _ in range(n):
        u = min((k, i) for i, k in enumerate(key) if not inMST[i])[1]
        inMST[u] = True
        for v in range(n):
            if matrix[u][v] > 0 and not inMST[v] and matrix[u][v] < key[v]:
                key[v] = matrix[u][v]
    return sum(key)

# Generating a dense graph with 100 vertices
n = 100
matrix = [[0 if i == j else (i + j) % 10 + 1 for j in range(n)] for i in range(n)]

cost = prim_matrix(n, matrix)

print("--- Practice Problem 3c: Dense Graph Matrix Prim's ---")
print(f"Number of Vertices: {n}")
print(f"Adjacency Matrix Prim's MST Total Cost: {cost}")
```
