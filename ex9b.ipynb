```python
def best_fit_decreasing(tasks, capacity=8.0):
    sorted_tasks = sorted(tasks, reverse=True)
    bins = []
    for task in sorted_tasks:
        best_idx, min_space = -1, float('inf')
        for i, space in enumerate(bins):
            if space >= task and space - task < min_space:
                min_space = space - task
                best_idx = i
        if best_idx >= 0:
            bins[best_idx] -= task
        else:
            bins.append(capacity - task)
    return len(bins)

tasks = [1.5, 2.0, 3.5, 0.5, 4.0, 1.0, 2.5, 3.0, 1.5, 2.0, 0.5, 4.0, 1.0, 3.5, 2.0]
servers_needed = best_fit_decreasing(tasks, 8.0)

print("--- Practice Problem 9b: Cloud Resource Allocation ---")
print(f"Total Tasks: {len(tasks)}")
print(f"Total Servers Needed (8GB RAM capacity): {servers_needed}")
```
