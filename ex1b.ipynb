```python
import random

def interpolation_search(arr, target):
    low, high = 0, len(arr) - 1
    probes = 0
    while low <= high and arr[low] <= target <= arr[high]:
        probes += 1
        if low == high:
            return (low, probes) if arr[low] == target else (-1, probes)
        pos = low + int(((target - arr[low]) * (high - low)) / (arr[high] - arr[low]))
        if arr[pos] == target:
            return pos, probes
        elif arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1
    return -1, probes

def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    probes = 0
    while low <= high:
        probes += 1
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid, probes
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, probes

# Roll numbers search demonstration (1 to 10000)
roll_numbers = sorted(random.sample(range(1, 10001), 5000))
target_roll = roll_numbers[2500]

idx_is, probes_is = interpolation_search(roll_numbers, target_roll)
idx_bs, probes_bs = binary_search(roll_numbers, target_roll)

print("--- Practice Problem 1b: Student Roll Numbers Search ---")
print(f"Target Roll Number to find: {target_roll}")
print(f"Interpolation Search -> Found at index {idx_is} in {probes_is} probes")
print(f"Binary Search        -> Found at index {idx_bs} in {probes_bs} probes")
```
